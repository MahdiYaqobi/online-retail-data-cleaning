# Online Retail Returns Analysis

This notebook loads and analyzes the `online_retail_returns.csv` dataset, performing initial exploration, inspecting data types, missing values, summary statistics, and return patterns.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
%matplotlib inline
```

```python
# Load the returns dataset
returns_path = Path("../data/processed/online_retail_returns.csv")
df_returns = pd.read_csv(returns_path)

# Convert InvoiceDate to datetime
df_returns["InvoiceDate"] = pd.to_datetime(df_returns["InvoiceDate"])

# Display basic info
df_returns.info()
```

```python
# View first few rows
df_returns.head(10)
```

```python
# Check for missing values
missing_values = df_returns.isnull().sum()
missing_percentage = (df_returns.isnull().sum() / len(df_returns)) * 100

missing_df = pd.DataFrame({
    "Missing Count": missing_values,
    "Missing Percentage (%)": missing_percentage
})
print(missing_df)
```

```python
# Statistical summary of numerical columns
df_returns.describe()
```

```python
# Initial EDA on returns: Top countries and StockCodes with most returns
print("Top 10 Countries with Returns:")
print(df_returns["Country"].value_counts().head(10))

print("\nTop 10 StockCodes with Most Returns:")
print(df_returns["StockCode"].value_counts().head(10))

print("\nQuantity summary in returns:")
print(df_returns["Quantity"].describe())

print("\nTotal Return Revenue impact:")
print(format(df_returns["TotalPrice"].sum(), ",.2f"))
```
